# **CST8508 Machine Vision - Lab 4: Streaming Live Webcam Video with Timestamp Overlay using OpenCV**

**Objective:**

By the end of this lab, you will learn how to access a live video stream from your webcam using the OpenCV library in Python and overlay the current time on the video feed.

**Requirements:**

*   A laptop with a working webcam.
*   Python installed on your laptop.
*   OpenCV library installed in your Python environment. You can install it using pip by running pip install opencv-python-headless in your terminal or command prompt.
*   Any text editor or IDE for writing Python code (e.g., VSCode, PyCharm). You can not do this lab on Google Colab


**Instructions:**

*   Ensure that Python and OpenCV are correctly installed on your laptop by running python --version and pip show opencv-python-headless in your terminal.
If you encounter any installation issues, please refer to the official OpenCV documentation or seek assistance.
Accessing the Webcam:

*   Start by importing the required OpenCV module in your Python script.
Initialize the webcam using OpenCV's VideoCapture object. The webcam is usually accessed using an index (0 for the default camera).
Create an infinite loop to continuously capture frames from the camera.
Displaying the Video Stream:

*   Within the loop, retrieve each frame from the webcam and display it in a window using OpenCV's imshow function.
Implement a mechanism to break out of the loop and close the window, typically by checking for a specific key press (e.g., the 'q' key).
Overlaying the Current Time:

*   In the same loop where you're displaying each frame, get the current time using Python's datetime module.
Use OpenCV's putText function to overlay the current time on each frame. You'll need to decide on the position, font, scale, and color of the text. Include milliseconds, timezone, and 24‑hour format for displaying time.
Cleanup:

*   After breaking out of the loop, release the webcam and close any OpenCV windows properly. This is crucial for freeing hardware resources and avoiding issues with subsequent runs of your script.
Testing:

*   Run your script to start streaming video from your webcam. The video should appear in a new window with the current time overlaid on each frame.
Test the key press functionality to ensure that the video stream can be terminated properly.
Troubleshooting:

*   If you encounter any issues with accessing the webcam or displaying the video, check your camera permissions and ensure that no other applications are using the camera.

*   For issues related to the timestamp overlay, verify the parameters used in the putText function for correctness and visibility.
