# 4. Classification (Peng Wang)

## 4.1 Data Preparation

### 4.1.1 Select Data

For classification, the target variable is the species class label. The Specimen Number is excluded as it is only an identifier. All 14 numerical features are retained initially. For images, only the 30 species present in both BW and feature datasets are used.

```python
# Select data for classification
X_cls = df[feature_cols].copy()
y_cls = df['Class'].copy()

print("Classification - Selected Data:")
print(f"  Features: {X_cls.shape}")
print(f"  Target: {y_cls.shape}, {y_cls.nunique()} classes")
print(f"  Features: {list(X_cls.columns)}")
```

### 4.1.2 Clean Data

No missing values, duplicates, or erroneous entries found. Class labels are encoded to consecutive integers (0–29) to avoid issues with certain algorithms.

```python
# Encode class labels
le_cls = LabelEncoder()
y_cls_enc = le_cls.fit_transform(y_cls)

mapping = pd.DataFrame({'Original': le_cls.classes_, 'Encoded': range(len(le_cls.classes_))})
print("Label Encoding:")
print(mapping.to_string(index=False))
print(f"\nEncoded range: {y_cls_enc.min()} to {y_cls_enc.max()}")
```

### 4.1.3 Construct Data

Five datasets are constructed as required:
1. Original data (unscaled)
2. Normalized data (Min-Max to [0, 1])
3. Standardized data (Z-score)
4. PCA-reduced standardized data (≥ 95% variance)
5. LDA-reduced standardized data

Additionally, a CNN model extracts features from the image dataset.

```python
# Construct 5 datasets for classification

# 1. Original
print("1. Original:", X_cls.shape)

# 2. Normalized
mm_cls = MinMaxScaler()
X_cls_norm = pd.DataFrame(mm_cls.fit_transform(X_cls), columns=feature_cols)
print(f"2. Normalized: {X_cls_norm.shape}, range [{X_cls_norm.min().min():.3f}, {X_cls_norm.max().max():.3f}]")

# 3. Standardized
ss_cls = StandardScaler()
X_cls_std = pd.DataFrame(ss_cls.fit_transform(X_cls), columns=feature_cols)
print(f"3. Standardized: {X_cls_std.shape}, mean~{X_cls_std.mean().mean():.4f}, std~{X_cls_std.std().mean():.4f}")

# 4. PCA-reduced
pca_cls = PCA(n_components=0.95)
X_cls_pca = pca_cls.fit_transform(X_cls_std)
print(f"4. PCA-reduced: {X_cls_pca.shape}, variance explained: {pca_cls.explained_variance_ratio_.sum():.4f}")

# 5. LDA-reduced
lda_cls = LDA()
X_cls_lda = lda_cls.fit_transform(X_cls_std, y_cls_enc)
print(f"5. LDA-reduced: {X_cls_lda.shape}")
```

```python
# Load and preprocess images for CNN feature extraction
IMG_SIZE = (128, 128)

def load_images(base_dir, target_classes, img_size=IMG_SIZE):
    images, labels = [], []
    for sp_folder in sorted(os.listdir(base_dir)):
        sp_path = os.path.join(base_dir, sp_folder)
        if not os.path.isdir(sp_path):
            continue
        class_num = int(sp_folder.split('.')[0].strip())
        if class_num not in target_classes:
            continue
        for img_file in sorted(os.listdir(sp_path)):
            try:
                img = Image.open(os.path.join(sp_path, img_file)).convert('RGB')
                img = img.resize(img_size)
                img_arr = np.array(img) / 255.0
                images.append(img_arr)
                labels.append(class_num)
            except Exception as e:
                print(f"Error: {e}")
    return np.array(images), np.array(labels)

target_classes = sorted(df['Class'].unique())
X_imgs, y_imgs = load_images(RGB_DIR, target_classes)

le_img = LabelEncoder()
y_imgs_enc = le_img.fit_transform(y_imgs)

print(f"Images loaded: {X_imgs.shape}")
print(f"Classes: {len(np.unique(y_imgs_enc))}")
```

```python
# CNN model for feature extraction (PyTorch)
class LeafCNN(nn.Module):
    def __init__(self, num_classes):
        super(LeafCNN, self).__init__()
        # Convolution Block 1
        self.conv1 = nn.Conv2d(3, 32, kernel_size=3, padding=1)
        self.pool1 = nn.MaxPool2d(2, 2)
        self.drop1 = nn.Dropout2d(0.25)

        # Convolution Block 2
        self.conv2 = nn.Conv2d(32, 64, kernel_size=3, padding=1)
        self.pool2 = nn.MaxPool2d(2, 2)
        self.drop2 = nn.Dropout2d(0.25)

        # Convolution Block 3
        self.conv3 = nn.Conv2d(64, 128, kernel_size=3, padding=1)
        self.pool3 = nn.MaxPool2d(2, 2)
        self.drop3 = nn.Dropout2d(0.25)

        # Fully connected
        self.flatten = nn.Flatten()
        self.fc1 = nn.Linear(128 * 16 * 16, 256)
        self.drop_fc = nn.Dropout(0.5)
        self.fc2 = nn.Linear(256, num_classes)

    def forward(self, x):
        x = self.drop1(self.pool1(F.relu(self.conv1(x))))
        x = self.drop2(self.pool2(F.relu(self.conv2(x))))
        x = self.drop3(self.pool3(F.relu(self.conv3(x))))
        x = self.flatten(x)
        x = self.drop_fc(F.relu(self.fc1(x)))
        x = self.fc2(x)
        return x

    def extract_features(self, x):
        """Extract features from the fc1 layer (256-dim vector)."""
        x = self.drop1(self.pool1(F.relu(self.conv1(x))))
        x = self.drop2(self.pool2(F.relu(self.conv2(x))))
        x = self.drop3(self.pool3(F.relu(self.conv3(x))))
        x = self.flatten(x)
        x = F.relu(self.fc1(x))
        return x

num_classes = len(np.unique(y_imgs_enc))
cnn_model = LeafCNN(num_classes).to(device)
print(cnn_model)
total_params = sum(p.numel() for p in cnn_model.parameters())
print(f"\nTotal parameters: {total_params:,}")
```

## 4.2 Modelling

### 4.2.1 Select Modeling Techniques

Four supervised classification algorithms are selected, plus the custom CNN:

1. **Gaussian Naïve Bayes** – Simple probabilistic baseline
2. **Support Vector Machine (SVM)** – Effective for high-dimensional data with RBF kernel
3. **Multi-Layer Perceptron (MLP)** – Neural network with hidden layers
4. **Stacking Classifier** – Ensemble that combines the above three as base estimators with Logistic Regression as the final estimator
5. **CNN** – Convolutional neural network trained on leaf images (feature-extraction approach from Section 4.1)

### 4.2.2 Generate Test Design

Each model is trained on each of the six prepared datasets (Original, Normalized, Standardized, PCA, LDA, CNN-extracted). Evaluation uses the 80/20 stratified split from Section 4.1.5. Metrics: Accuracy, Precision (macro), Recall (macro), F1 (macro).

```python
# 4.2.3 Build model  &  4.2.4 Assess model
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
from sklearn.neural_network import MLPClassifier
from sklearn.ensemble import StackingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report, confusion_matrix
import warnings
warnings.filterwarnings('ignore')

# Define classifiers
cls_models = {
    'GaussianNB': GaussianNB(),
    'SVM': SVC(kernel='rbf', random_state=42),
    'MLP': MLPClassifier(hidden_layer_sizes=(128, 64), max_iter=500, random_state=42),
    'Stacking': StackingClassifier(
        estimators=[
            ('nb', GaussianNB()),
            ('svm', SVC(kernel='rbf', random_state=42)),
            ('mlp', MLPClassifier(hidden_layer_sizes=(128, 64), max_iter=500, random_state=42))
        ],
        final_estimator=LogisticRegression(max_iter=1000, random_state=42),
        cv=5
    )
}

# Datasets (feature-based)
cls_datasets = {
    'Original': (X_tr_orig, X_te_orig, y_tr, y_te),
    'Normalized': (X_tr_norm, X_te_norm, y_tr, y_te),
    'Standardized': (X_tr_std, X_te_std, y_tr, y_te),
    'PCA': (X_tr_pca, X_te_pca, y_tr, y_te),
    'LDA': (X_tr_lda, X_te_lda, y_tr, y_te),
    'CNN-features': (X_tr_cnn, X_te_cnn, y_tr_img, y_te_img)
}

cls_results = []

for ds_name, (Xtr, Xte, ytr, yte) in cls_datasets.items():
    for model_name, model in cls_models.items():
        import copy
        clf = copy.deepcopy(model)
        clf.fit(Xtr, ytr)
        y_pred = clf.predict(Xte)
        acc  = accuracy_score(yte, y_pred)
        prec = precision_score(yte, y_pred, average='macro', zero_division=0)
        rec  = recall_score(yte, y_pred, average='macro', zero_division=0)
        f1   = f1_score(yte, y_pred, average='macro', zero_division=0)
        cls_results.append({
            'Dataset': ds_name,
            'Model': model_name,
            'Accuracy': round(acc, 4),
            'Precision': round(prec, 4),
            'Recall': round(rec, 4),
            'F1': round(f1, 4)
        })
        print(f"[Classification] {ds_name:20s} | {model_name:12s} | Acc={acc:.4f} P={prec:.4f} R={rec:.4f} F1={f1:.4f}")

cls_results_df = pd.DataFrame(cls_results)
print("\n", cls_results_df.to_string(index=False))
```

## 5.2 Modelling

### 5.2.1 Select Modeling Techniques

Three clustering algorithms are selected:

1. **K-Means** (k = 30, matching the known number of species)
2. **DBSCAN** – Density-based clustering
3. **Gaussian Mixture (EM)** – Expectation-Maximization with 30 components

### 5.2.2 Generate Test Design

Clustering uses the full dataset (no train-test split). Performance is measured using external validation metrics: **Adjusted Rand Index (ARI)** and **Normalized Mutual Information (NMI)**, comparing predicted clusters against the true species labels.

```python
# 5.2.3 Build model  &  5.2.4 Assess model
from sklearn.cluster import KMeans, DBSCAN
from sklearn.mixture import GaussianMixture
from sklearn.metrics import adjusted_rand_score, normalized_mutual_info_score, silhouette_score

n_species = 30

clu_models = {
    'KMeans': KMeans(n_clusters=n_species, random_state=42, n_init=10),
    'DBSCAN': DBSCAN(eps=2.0, min_samples=3),
    'GaussianMixture': GaussianMixture(n_components=n_species, random_state=42)
}

clu_datasets = {
    'Original': X_clu,
    'Normalized': X_clu_norm,
    'Standardized': X_clu_std,
    'PCA': X_clu_pca,
    'LDA': X_clu_lda,
    'CNN-features': X_cnn_features
}

clu_results = []

for ds_name, X_data in clu_datasets.items():
    for model_name, model in clu_models.items():
        import copy
        clust = copy.deepcopy(model)
        if model_name == 'GaussianMixture':
            clust.fit(X_data)
            labels = clust.predict(X_data)
        else:
            labels = clust.fit_predict(X_data)
        # Skip metrics if only one cluster or all noise
        n_labels = len(set(labels) - {-1})
        if n_labels < 2:
            ari, nmi, sil = -1, -1, -1
        else:
            ari = adjusted_rand_score(y_clu_enc, labels)
            nmi = normalized_mutual_info_score(y_clu_enc, labels)
            try:
                sil = silhouette_score(X_data, labels)
            except:
                sil = -1
        clu_results.append({
            'Dataset': ds_name,
            'Model': model_name,
            'Clusters': n_labels,
            'ARI': round(ari, 4),
            'NMI': round(nmi, 4),
            'Silhouette': round(sil, 4)
        })
        print(f"[Clustering] {ds_name:20s} | {model_name:18s} | Clusters={n_labels:3d} ARI={ari:.4f} NMI={nmi:.4f} Sil={sil:.4f}")

clu_results_df = pd.DataFrame(clu_results)
print("\n", clu_results_df.to_string(index=False))
```

## 5.3 Evaluation

### 5.3.1 Evaluate Results

The table above summarises clustering performance across all dataset–model combinations.

Standardized and LDA-reduced data tend to produce the best clustering results, consistent with the fact that distance-based methods benefit from normalised feature scales.

## 6.2 Modelling

### 6.2.1 Select Modeling Techniques

Three outlier detection algorithms are selected:

1. **Local Outlier Factor (LOF)** – Density-based local outlier detection
2. **Isolation Forest** – Tree-based anomaly detection
3. **One-Class SVM** – Boundary-based novelty detection

A contamination parameter of 0.1 is used (assuming approximately 10% of specimens are unusual).

### 6.2.2 Generate Test Design

Outlier detection uses the full dataset. Since no ground-truth outlier labels exist, evaluation is based on:
- Number and percentage of detected outliers
- Agreement across methods
- Distribution analysis of detected outliers across species

```python
# 6.2.3 Build model  &  6.2.4 Assess model
from sklearn.neighbors import LocalOutlierFactor
from sklearn.ensemble import IsolationForest
from sklearn.svm import OneClassSVM

contamination = 0.1

out_models = {
    'LOF': LocalOutlierFactor(n_neighbors=20, contamination=contamination),
    'IsolationForest': IsolationForest(contamination=contamination, random_state=42),
    'OneClassSVM': OneClassSVM(nu=contamination, kernel='rbf', gamma='scale')
}

out_datasets = {
    'Original': X_out,
    'Normalized': X_out_norm,
    'Standardized': X_out_std,
    'PCA': X_out_pca,
    'LDA': X_out_lda
}

out_results = []

for ds_name, X_data in out_datasets.items():
    for model_name, model in out_models.items():
        import copy
        det = copy.deepcopy(model)
        if model_name == 'LOF':
            preds = det.fit_predict(X_data)
        else:
            det.fit(X_data)
            preds = det.predict(X_data)
        # preds: 1 = inlier, -1 = outlier
        n_outliers = (preds == -1).sum()
        pct_outliers = n_outliers / len(preds) * 100
        out_results.append({
            'Dataset': ds_name,
            'Model': model_name,
            'Outliers': n_outliers,
            'Pct_Outliers': round(pct_outliers, 2)
        })
        print(f"[Outlier] {ds_name:20s} | {model_name:18s} | Outliers={n_outliers:3d} ({pct_outliers:.1f}%)")

out_results_df = pd.DataFrame(out_results)
print("\n", out_results_df.to_string(index=False))
```

### 6.3.2 Interpret Results
